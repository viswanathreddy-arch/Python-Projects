```python
df = df[["date", "location", "total_cases", "new_cases", "total_deaths", "new_deaths", "total_vaccinations"]]
```

```python
df = df.sort_values(["location", "date"])
```

```python
cum_cols = ["total_cases",  "total_deaths", "total_vaccinations"]
df[cum_cols] = df.groupby("location")[cum_cols].ffill()
```

```python
df = df.fillna(0)
```

```python
df.isnull().sum()
```

```python
pivot_table = pd.pivot_table(
    data=df,
    index="date",
    columns="location",
    values="new_cases",
)
```

```python
total = df.groupby("location")[["total_cases", "total_deaths"]].max()
total
```

```python
death_rate = (df.groupby("location")[["total_deaths", "total_cases"]].max())

death_rate["Death_rate"] = (
    death_rate["total_deaths"] /
    death_rate["total_cases"] * 100
).round(2)
```

```python
death_rate
```

```python
death_rate = death_rate.sort_values(by="Death_rate", ascending=False)
```

```python
death_rate
```

```python
first_month = df.groupby("location").head(90)
last_month = df.groupby("location").tail(90)
```

```python
growth = (
    first_month.groupby("location")["new_cases"].mean() /
    last_month.groupby("location")["new_cases"].mean()
).sort_values(ascending=False)

growth.idxmax()
```
